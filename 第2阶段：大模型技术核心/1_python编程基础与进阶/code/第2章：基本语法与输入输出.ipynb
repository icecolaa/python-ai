### 1. 代码规范与注释——写出“好看”的代码

Python 强制使用缩进表示代码块，省去了 C/Java 风格的大括号，使程序结构清晰。

#### 1.1 基本行结构
- Python 语句以**换行结束**，无需分号 `;`。
- 若一行太长，可用反斜杠 `\` 续行，或在圆括号、方括号、花括号内直接换行。

```python
# 正常换行
print("第一行")
print("第二行")

# 续行符
total = 1 + 2 + 3 + \
        4 + 5 + 6
print(total)

# 括号内直接换行（推荐，更美观）
fruits = ["苹果", "香蕉",
          "橘子", "葡萄"]
print(fruits)
```

#### 1.2 缩进规则（核心）
- Python **用缩进表示代码块**，同级代码必须保持相同缩进。
- 推荐使用 **4 个空格** 或 1 个 Tab（整个文件保持统一，PEP 8 建议用空格）。
- 缩进错误会引发 `IndentationError`。

```python
# ✅ 正确
if True:
    print("缩进了 4 个空格")
    print("同属一个代码块")

# ❌ 错误（缺少缩进）
if True:
print("这行会报错")   # IndentationError: expected an indented block
```

#### 1.3 注释
- **单行注释**：以 `#` 开头，`#` 后建议加一个空格。
- **多行注释/文档字符串**：用三个单引号 `'''` 或三个双引号 `"""` 包裹，通常用于函数、类、模块的说明。

```python
name = "Alice"      # 这是行尾注释，说明变量含义

def add(a, b):
    """
    两数相加函数
    参数 a: 第一个加数
    参数 b: 第二个加数
    返回: 两数和
    """
    return a + b

content = '''
零基础 学AI 找白泽 不踩坑
白泽是专业的IT团队
欢迎大家的到来
'''
print(content)
```

### 2. 变量与标识符——给数据贴标签

#### 2.1 变量：动态、三要素、无需声明
Python 变量是动态类型的：**赋值时就决定了类型，且之后可以改变类型**。  
每个变量包含三个要素：
- **id**：内存地址（唯一标识）
- **type**：数据类型
- **value**：具体值

```python
x = 10
print(id(x), type(x), x)   # 输出类似: 140736... <class 'int'> 10

x = 3.14
print(id(x), type(x), x)   # id 改变，类型变为 float
```

#### 2.2 标识符命名规范
- 只能由字母（区分大小写）、数字、下划线 `_` 组成。
- 不能以数字开头。
- 不能与 Python **关键字**（如 `if`, `for`, `while`）重名。
- 建议：**见名知意**，使用下划线命名法 `my_var`(更推荐) 或驼峰命名法 `myVar`。

```python
# 查看所有关键字
import keyword
print(keyword.kwlist)
print(f"关键字总数: {len(keyword.kwlist)}")

# 判断是否为关键字
print(keyword.iskeyword('for'))   # True
print(keyword.iskeyword('end'))   # False
```

#### 2.3 标识符示例

```python
student_name = "小明"   # 下划线命名
studentAge = 20         # 驼峰命名（也可用，但 PEP8 推荐下划线用于变量）
# 2name = "error"       # 错误：不能以数字开头
# class = "error"       # 错误：class 是关键字
```

### 3. 输入与输出——让程序与你对话

| 操作 | 语法                                   | 说明                                         |
| ---- | -------------------------------------- | -------------------------------------------- |
| 输出 | `print(value, ..., sep=' ', end='\n')` | `sep` 分隔符默认空格，`end` 结尾符默认换行   |
| 输入 | `variable = input("提示信息")`         | **接收的数据类型始终是字符串**，计算前需转换 |

#### 3.1 基本输出

```python
print("HelloWorld")
print("Hello", "World")                # Hello World
print("Hello", "World", sep="-")       # Hello-World
print("Hello", end="")                 # 不换行
print("World")                         # HelloWorld
```

#### 3.2 格式化输出（从旧到新）
- **% 格式化**（旧风格，了解即可）
- **str.format()**（灵活但稍显冗长）
- **f-string**（Python 3.6+ **强烈推荐**，简洁高效，支持表达式）

```python
name = "Alice"
age = 25

# 1. % 格式化
# %s %f %d %i %c %o #x %.2f
print("姓名: %s, 年龄: %d" % (name, age))
print("姓名: %s, 年龄: %.2f" % (name, age))

# 2. format() 方法
print("姓名: {0}, 年龄: {1}".format(name, age))
print("姓名: {0}, 年龄: {1}".format(age, name))
print("姓名: {name}, 年龄: {age:.2f}".format(name=name, age=age))
print("姓名: {name}, 年龄: {age}".format(age=age, name=name))
print("{:*^20}".format("居中填充"))   # ******居中填充*******

# 3. f-string （推荐）
print(f"姓名: {name}, 年龄: {age}")
print(f"{3.14159:.2f}")              # 3.14  控制浮点数精度
print(f"{age:>10}")                  # 右对齐，宽度 10
```

#### 3.3 输入与类型转换
`input()` 返回的是字符串，数学运算前必须转为数字。

```python
# ❌ 错误示例：字符串相加变成了拼接
a = input("请输入第一个整数: ")   # 输入 10
b = input("请输入第二个整数: ")   # 输入 20
c = a + b
print(f"{a} + {b} = {c}")         # 10 + 20 = 1020

# ✅ 正确做法：使用 int() 或 float() 转换
a = int(input("请输入第一个整数: "))
b = int(input("请输入第二个整数: "))
c = a + b
print(f"{a} + {b} = {c}")         # 10 + 20 = 30
```

#### 3.4 转义字符速查
- `\n` 换行
- `\t` 制表符（Tab）
- `\\` 反斜杠本身
- `\"` 双引号

```python
print("第一行\n第二行")            # 两行输出
print("姓名\t年龄\t城市")          # 各列对齐
print("0基础 学AI 找\"白泽\" 不踩坑")   # 转义双引号
```

### 4. 运算符——赋予程序计算能力

#### 4.1 算术运算符

```python
print(10 + 3)    # 13
print(10 - 3)    # 7
print(10 * 3)    # 30
print(10 / 3)    # 3.333...  真除，结果为浮点数
print(10 // 3)   # 3         整除（向下取整）
print(10 % 3)    # 1         取模（求余）
print(2 ** 3)    # 8         幂
```

#### 4.2 赋值运算符与复合赋值

```python
x = 10
x += 5      # 等价于 x = x + 5
x *= 2      # 等价于 x = x * 2
print(x)    # 30
x /= 3      # 等价于 x = x / 3
print(x)    # 10.0
x **= 2     # 等价于 x = x ** 2
print(x)    # 100.0
```

**经典例题**：

```python
a = 12
a += a * a   # a = a + (a*a) = 12 + 144 = 156
print(a)     # 156

i, j, k = 10, 10, 3
k *= i + j   # k = k * (i+j) = 3 * 20 = 60
print(k)     # 60
```

#### 4.3 比较运算符（返回布尔值）

```python
print(3 > 2)   # True
print(3 < 2)   # False
print(3 == 3)  # True
print(3 != 3)  # False
print(3 >= 2)  # True
print(3 <= 3)  # True
```

**常用判断模板**（务必熟练）：
- 判断偶数：`num % 2 == 0`
- 判断奇数：`num % 2 != 0`
- 判断5的倍数：`num % 5 == 0`

#### 4.4 逻辑运算符：`and`, `or`, `not`
- `and`：两边都为 `True` 才为 `True`，全真才真，遇假则假
- `or`：至少一边为 `True` 就为 `True`，遇真则真，全假才假
- `not`：取反，非真即假，非假即真

```python
print(3 > 2 and 5 > 3)   # True
print(3 > 5 or 5 > 3)    # True
print(not 3 > 2)         # False
```

**注意**：布尔值可参与整数运算，`True` 为 1，`False` 为 0。

```python
x = 2
print(x + (3 > 2))   # 2 + True = 2 + 1 = 3
```

```python
# 补充逻辑短路现象 （以除数不能为0的语法错误进行示例）
# print(3>2 and 3/0)  # ZeroDivisionError: division by zero
print(3<2 and 3/0)  # False
# print(3<2 or 3/0)  # ZeroDivisionError: division by zero
print(3>2 or 3/0)   # True
```

**经典条件组合**：
- 能被3或5整除但不能被7整除：
  `(num % 3 == 0 or num % 5 == 0) and num % 7 != 0`
- 判断三位数：
  `100 <= num <= 999` 或 `num > 99 and num < 1000`
- 闰年判断：
  `(year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)`

#### 4.5 位运算符（二进制层操作）
直接对整数的二进制补码按位运算：
| 运算符 | 含义                     | 示例       | 结果 |
| ------ | ------------------------ | ---------- | ---- |
| `&`    | 按位与                   | `25 & 13`  | `9`  |
| `\|`   | 按位或                   | `25 \| 13` | `29` |
| `^`    | 按位异或                 | `25 ^ 13`  | `20` |
| `~`    | 按位取反                 | `~(-3)`    | `2`  |
| `<<`   | 左移（低位补0）          | `3 << 1`   | `6`  |
| `>>`   | 右移（正数补0，负数补1） | `3 >> 1`   | `1`  |

**常见题型**（十六进制/八进制参与位运算）：

```python
# 0x10 | 0x01 = 0x11
# 0x10 & 0x01 = 0x00
# 12 & 0o12 的十进制结果：12 & 10 = 8
```

#### 4.6 运算符优先级（由高到低）
1. `()` 括号
2. `**` 乘方
3. `*`, `/`, `//`, `%` 乘除
4. `+`, `-` 加减
5. 比较运算符（`>`, `<`, `==` 等）
6. `not`, `and`, `or`
7. `=` 赋值

> **技巧**：不确定优先级时直接加括号，增强可读性。

### 5. 数据类型转换——打通字符串与数字的壁垒

前面看到 `input()` 返回字符串，需转换为数字才能运算。Python 提供了一组内置转换函数：

| 函数       | 作用            | 示例                     |
| ---------- | --------------- | ------------------------ |
| `int(x)`   | 转为整数        | `int("123")` → `123`     |
| `float(x)` | 转为浮点数      | `float("3.14")` → `3.14` |
| `str(x)`   | 转为字符串      | `str(100)` → `"100"`     |
| `bool(x)`  | 转为布尔值      | `bool(0)` → `False`      |
| `chr(x)`   | ASCII 码→字符   | `chr(65)` → `'A'`        |
| `ord(x)`   | 字符→ASCII 码   | `ord('a')` → `97`        |
| `bin(x)`   | 十进制→二进制   | `bin(10)` → `'0b1010'`   |
| `hex(x)`   | 十进制→十六进制 | `hex(255)` → `'0xff'`    |
| `oct(x)`   | 十进制→八进制   | `oct(8)` → `'0o10'`      |

**应用演示**：大小写字母转换原理（ASCII 码相差 32）

```python
ch = 'a'
upper = chr(ord(ch) - 32)
print(f"{ch} 的大写是 {upper}")   # A
```

```python
num = 123
print(id(num), type(num), num)
print(num + 456)
num = str(num)
print(id(num), type(num), num)
print(num + "456")
```

```python
# 在实际的计算中，存在隐式类型转换
a = 1
b = 1.5
c = a + b
print(f"{a} + {b} = {c}")
```

### 6. 学以致用：简易计算器（串联全部知识点）

需求：用户输入两个数字，程序计算它们的加、减、乘、除并格式化输出结果，同时判断两数是否相等。

```python
# 简易计算器 - 演示变量、输入输出、运算符与类型转换
print("=== 简易计算器 ===")
num1 = float(input("请输入第一个数字: "))
num2 = float(input("请输入第二个数字: "))

# 计算
add = num1 + num2
sub = num1 - num2
mul = num1 * num2
div = num1 / num2
eq = num1 == num2

# 格式化输出
print(f"\n{num1} + {num2} = {add}")
print(f"{num1} - {num2} = {sub}")
print(f"{num1} × {num2} = {mul}")
print(f"{num1} ÷ {num2} = {div}")
print(f"两数相等吗？ {eq}")
```

### 7. 练习

**练习 1（水仙花数判断）**  
输入一个三位数，判断它是否为水仙花数（各位数字立方和等于本身，如 153=1³+5³+3³）。

```python
num = int(input("请输入一个三位数："))
g = num % 10
s = num // 10 % 10
b = num // 100 % 10
print(f"输入的三位数 {num} 是水仙花数吗？{g ** 3 + s ** 3 + b ** 3 == num}")
```
